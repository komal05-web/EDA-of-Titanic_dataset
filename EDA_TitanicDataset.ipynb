```python
import pandas as pd
import numpy as np
df=pd.read_csv(r"C:\Users\Komal Pandey\Downloads\train.csv")
df.head()
```

```python
# Check dataset structure
print(df.info())
```

```python
# Check summary statistics
print(df.describe())
```

```python
# Check missing values
print(df.isnull().sum())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Survival rate by gender
sns.barplot(x='Sex', y='Survived', data=df)
plt.title('Survival Rate by Gender')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Select numerical features for pairplot
numerical_features = ['Age', 'Fare', 'Pclass', 'SibSp', 'Parch']

# Create pairplot
sns.pairplot(df[numerical_features], diag_kind='kde', corner=True)
plt.suptitle('Pairplot of Numerical Features', y=1.02)
plt.show()
```

```python
# Compute correlation matrix
correlation_matrix = df[numerical_features].corr()

# Create heatmap
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Plot histogram for Age
sns.histplot(df['Age'], bins=20, kde=True, color='blue')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
# Boxplot for Fare by Pclass
sns.boxplot(x='Pclass', y='Fare', hue='Pclass', data=df, palette='pastel', legend=False)
plt.title('Fare Distribution Across Passenger Classes')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.show()
```

```python
# Scatterplot for Age vs Fare
sns.scatterplot(x='Age', y='Fare', hue='Survived', data=df, palette='Set1', alpha=0.7)
plt.title('Scatterplot of Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.legend(title='Survived')
plt.show()
```

#Conclusion:The visualizations give us a better understanding of passenger demographics, financial capacity, and their connection to survival. For example, survival is influenced by factors like Pclass, Fare, and potentially age. These insights can be valuable for predictive modeling (e.g., identifying which factors most impact survival probability).
